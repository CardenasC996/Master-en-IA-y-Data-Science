```python
#Notas de estudiante:

nota = 4.5
umbral = 3.6

# Si la nota es mayor o igual al umbral, el estudiante aprueba, de lo contrario, reprueba condicional if-else
# palabra reservada if, seguida de la condicion a evaluar, seguida de dos puntos, luego el bloque de codigo a ejecutar si la condicion es verdadera, luego la palabra reservada else, seguida de dos puntos, luego el bloque de codigo a ejecutar si la condicion es falsa
if nota >= umbral:
    resultado = "Aprobado"
else:
    resultado = "Reprobado" 

print(f"ejemplo 1  - Nota: {nota} => {resultado}")
```

```python
nota = 4.6

if nota > 4.5:
    resultado = "Excelente"
elif nota >= 4:
    resultado = "Aprobado"
elif nota >= 3.5:    #si aqui colocamos un if, el programa se detendria al encontrar la primera condicion verdadera, por eso se utiliza elif para evaluar las siguientes condiciones
    resultado = "Regular"
else:
    resultado = "Reprobado"

print(f"ejemplo 2  - Nota: {nota} => {resultado}")
```
